```python
df.drop(columns='Best Seller',inplace=True)
```

```python
(df.isnull().mean() * 100).round(2)
```

```python
sns.boxplot(df['Dining Rating'])
```

```python
df['Dining Rating'] = df['Dining Rating'].fillna(df['Dining Rating'].median())
```

```python
sns.boxplot(df['Delivery Rating'])
```

```python
df['Delivery Rating'] = df['Delivery Rating'].fillna(df['Delivery Rating'].median())
```

```python
(df.isnull().mean() * 100).round(2)
```

```python
top10 = (df.groupby('Restaurant Name')['Dining Rating']
         .mean()
         .sort_values(ascending=False)
         .head(10))

plt.figure(figsize=(12,6))
sns.barplot(x=top10.values, y=top10.index)

plt.title('Top 10 Restaurants by Dining Rating')
plt.xlabel('Average Dining Rating')
plt.ylabel('Restaurant Name')
plt.show()
```

```python
cuisine = df['Cuisine'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=cuisine.values, y=cuisine.index)

plt.title('Top 10 Popular Cuisines')
plt.xlabel('Number of Records')
plt.ylabel('Cuisine')
plt.show()
```

```python
city = df.groupby('City')['Restaurant Name'].nunique()
city = city.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
city.plot(kind='bar', figsize=(10,6))

plt.title('Top 10 Cities by Restaurant Count')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.show()
```

```python
rating = (df.groupby('Cuisine')['Dining Rating']
          .mean()
          .sort_values(ascending=False)
          .head(10))

plt.figure(figsize=(10,6))
sns.barplot(x=rating.values, y=rating.index)

plt.title('Top 10 Cuisines by Average Rating')
plt.xlabel('Average Dining Rating')
plt.show()
```

```python
items = (df.groupby('Item Name')['Votes']
         .sum()
         .sort_values(ascending=False)
         .head(10))

plt.figure(figsize=(12,6))
sns.barplot(x=items.values, y=items.index)

plt.title('Top 10 Food Items by Votes')
plt.xlabel('Total Votes')
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df['Prices'].dropna(), bins=20, kde=True)

plt.title('Food Price Distribution')
plt.xlabel('Food Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='Dining Rating',
    y='Dining Votes'
)

plt.title('Dining Rating vs Dining Votes')
plt.xlabel('Dining Rating')
plt.ylabel('Dining Votes')
plt.show()
```

```python
plt.figure(figsize=(10,6))

corr = df[['Dining Rating',
           'Delivery Rating',
           'Dining Votes',
           'Delivery Votes',
           'Votes',
           'Prices']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm',
            fmt='.2f')

plt.title('Restaurant Data Correlation Heatmap')
plt.show()
```

```python
cuisine_price = (
    df.groupby('Cuisine')['Prices']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(x=cuisine_price.values,
            y=cuisine_price.index)

plt.title('Top 10 Cuisines by Average Food Price')
plt.xlabel('Average Price')
plt.ylabel('Cuisine')

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(data=df,
                x='Dining Rating',
                y='Delivery Rating',
                alpha=0.6)

plt.title('Dining Rating vs Delivery Rating')
plt.xlabel('Dining Rating')
plt.ylabel('Delivery Rating')

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16,12))

# Graph 1: Popular cuisines
df['Cuisine'].value_counts().head(10).plot(
    kind='bar', ax=axes[0,0]
)
axes[0,0].set_title('Top 10 Popular Cuisines')

# Graph 2: Price distribution
sns.histplot(df['Prices'].dropna(),
             kde=True, ax=axes[0,1])
axes[0,1].set_title('Food Price Distribution')

# Graph 3: Dining rating
sns.histplot(df['Dining Rating'].dropna(),
             bins=15, ax=axes[1,0])
axes[1,0].set_title('Dining Rating Distribution')

# Graph 4: Delivery rating
sns.histplot(df['Delivery Rating'].dropna(),
             bins=15, ax=axes[1,1])
axes[1,1].set_title('Delivery Rating Distribution')

plt.tight_layout()
plt.show()
```

```python
# Calculate KPIs

total_restaurants = df['Restaurant Name'].nunique()

total_cities = df['City'].nunique()

total_cuisines = df['Cuisine'].nunique()

total_food_items = df['Item Name'].nunique()

avg_dining_rating = df['Dining Rating'].mean()

avg_delivery_rating = df['Delivery Rating'].mean()

total_dining_votes = df['Dining Votes'].sum()

total_delivery_votes = df['Delivery Votes'].sum()

total_food_votes = df['Votes'].sum()

avg_food_price = df['Prices'].mean()

highest_food_price = df['Prices'].max()


# Display KPIs

print("TOTAL RESTAURANTS:", total_restaurants)
print("TOTAL CITIES:", total_cities)
print("TOTAL CUISINES:", total_cuisines)
print("TOTAL FOOD ITEMS:", total_food_items)
print("AVERAGE DINING RATING:", round(avg_dining_rating, 2))
print("AVERAGE DELIVERY RATING:", round(avg_delivery_rating, 2))
print("TOTAL DINING VOTES:", total_dining_votes)
print("TOTAL DELIVERY VOTES:", total_delivery_votes)
print("TOTAL FOOD VOTES:", total_food_votes)
print("AVERAGE FOOD PRICE:", round(avg_food_price, 2))
print("HIGHEST FOOD PRICE:", highest_food_price)
```
